```python
import numpy as np
import pandas as pd
import pickle as pkl
```

```python
similarity = pkl.load(open("similarity.pkl", "rb+"))
```

```python
movies = pd.read_csv("cleaned_data.csv")
```

```python
def get_movie_name(index):
    if index > len(movies) - 1:
        return ""
    else:
        return movies.iloc[index]["title"]
```

```python
def get_movie_index(name):
    index = -1
    for i in movies.index:
        if movies.iloc[i]["title"].lower() == name.lower():
            index = i
            break
    return index
```

```python
name = input("Enter movie name:")
index = get_movie_index(name)
if index == -1:
    print("Sorry, movie not found")
else:
    print("Movie Found\nNext Recommended Movies:")
    similarity_indexes = list(enumerate(similarity[index]))
    similarity_indexes = sorted(similarity_indexes, key=lambda x: x[1], reverse=True)    
    for i in range(1, 6):
        print(str(i) + ". " + get_movie_name(similarity_indexes[i][0]))
```
