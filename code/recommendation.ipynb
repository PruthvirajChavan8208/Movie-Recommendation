```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
ds = pd.read_csv("cleaned_data.csv")
```

```python
#Count no of tags for every movie
for i in ds.index:
    print(ds.iloc[i]["title"], "-", ds.iloc[i]["tags"].count(','))
```

```python
#Find 10000 unique words from all tags and create vector for every movie
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=10000,stop_words='english')
vector = cv.fit_transform(ds['tags']).toarray()
```

```python
# for word in cv.get_feature_names_out():
#     print(word)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
def get_movie_index(name):
    index = -1
    for i in ds.index:
        if ds.iloc[i]["title"].lower() == name.lower():
            index = i
            break
    return index
```

```python
def get_movie_name(index):
    if index > len(ds) - 1:
        return ""
    else:
        return ds.iloc[index]["title"]
```

```python
name = input("Enter movie name:")
index = get_movie_index(name)
if index == -1:
    print("Sorry, movie not found")
else:
    print("Movie Found\nNext Recommended Movies:")
    similarity_indexes = list(enumerate(similarity[index]))
    similarity_indexes = sorted(similarity_indexes, key=lambda x: x[1], reverse=True)    
    for i in range(1, 6):
        print(str(i) + ". " + get_movie_name(similarity_indexes[i][0]))
```

```python
import pickle as pkl
pkl.dump(similarity, open("similarity.pkl", "wb+"))
```
